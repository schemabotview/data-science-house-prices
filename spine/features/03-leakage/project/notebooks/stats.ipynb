```python
garage = homes["garage"] > 0
prices = homes["price"]
observed = prices[garage].median() - prices[~garage].median()
observed
```

```python
def shuffled_differences(prices, flags, n=5000, seed=1460):
    """Median difference between the flagged and the rest, with the flags shuffled."""
    rng = np.random.default_rng(seed)
    values, flags = np.asarray(prices), np.asarray(flags)
    out = np.empty(n)
    for i in range(n):
        f = rng.permutation(flags)
        out[i] = np.median(values[f]) - np.median(values[~f])
    return out

diffs = shuffled_differences(prices, garage)
(np.abs(diffs) >= abs(observed)).mean()
```

```python
ax = pd.Series(diffs).plot.hist(bins=40, title="5,000 shuffled differences")
ax.axvline(observed, color="red");
```

```python
round(observed / prices.median(), 3)
```

```python
few = homes.sample(40, random_state=7)
few_garage = few["garage"] > 0
few_prices = few["price"]
few_observed = few_prices[few_garage].median() - few_prices[~few_garage].median()
few_diffs = shuffled_differences(few["price"], few_garage)
few_observed, (np.abs(few_diffs) >= abs(few_observed)).mean()
```
