```python
import numpy as np
import pandas as pd

homes = pd.read_csv("../data/homes_clean.csv", parse_dates=["sold"])
homes["area"] = homes["area"].str.title()
areas = pd.read_csv("../data/areas.csv")
homes = homes.merge(areas, on="area", validate="many_to_one")
homes["price"].agg(["mean", "median", "skew"]).round(2)
```

```python
homes["price"].plot.hist(bins=range(0, 700_001, 25_000), title="price");
```

```python
log_price = np.log10(homes["price"])
log_price.plot.hist(bins=30, title="log10(price)");
```

```python
log_price.agg(["mean", "median", "skew"]).round(3)
```

```python
homes.plot.scatter(x="sqft", y="price", s=4, alpha=0.4, title="price against size");
```

```python
homes["sqft"].corr(homes["price"]).round(3)
```

```python
spread = homes.groupby("area")["price"].quantile([0.25, 0.5, 0.75]).unstack()
spread = spread.sort_values(0.5).round(-3)
spread
```

```python
ax = spread[0.5].plot.bar(title="median price, with the middle half", rot=0)
ax.vlines(range(len(spread)), spread[0.25], spread[0.75], color="black");
```

```python
numbers = ["price", "sqft", "beds", "built", "garage", "condition"]
numbers += ["school", "km_to_centre"]
homes[numbers].corr().round(2)
```

```python
homes[numbers].corr()["price"].drop("price").sort_values().round(2)
```
