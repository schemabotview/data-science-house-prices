```python
import numpy as np
import pandas as pd

homes = pd.read_csv("../data/homes_clean.csv", parse_dates=["sold"])
homes["area"] = homes["area"].str.title()
areas = pd.read_csv("../data/areas.csv")
homes = homes.merge(areas, on="area", validate="many_to_one")
homes["price"].agg(["mean", "median", "skew"]).round(2)
```

```python
homes["price"].plot.hist(bins=range(0, 700_001, 25_000), title="price");
```

```python
log_price = np.log10(homes["price"])
log_price.plot.hist(bins=30, title="log10(price)");
```

```python
log_price.agg(["mean", "median", "skew"]).round(3)
```
