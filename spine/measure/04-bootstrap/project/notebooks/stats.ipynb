```python
import numpy as np
import pandas as pd

homes = pd.read_csv("../data/homes_clean.csv", parse_dates=["sold"])
homes["area"] = homes["area"].str.title()
areas = pd.read_csv("../data/areas.csv")
homes = homes.merge(areas, on="area", validate="many_to_one")
homes["price"].agg(["mean", "median", "std"]).round()
```

```python
q1, q3 = homes["price"].quantile([0.25, 0.75])
q3 - q1
```

```python
plus_one = pd.concat([homes["price"], pd.Series([5_000_000])])
pd.DataFrame({
    "as sold": homes["price"].agg(["mean", "median", "std"]),
    "+ one £5m sale": plus_one.agg(["mean", "median", "std"]),
}).round()
```

```python
sample = homes["price"].sample(100, random_state=1)
sample.median(), homes["price"].median()
```

```python
medians = pd.Series(
    [homes["price"].sample(100, random_state=i).median() for i in range(1000)]
)
medians.describe().round()
```

```python
medians.plot.hist(bins=30, title="medians of 1,000 samples of 100");
```

```python
riverside = homes.loc[homes["area"] == "Riverside", "price"]
len(riverside), riverside.median()
```

```python
boot = pd.Series([
    riverside.sample(len(riverside), replace=True, random_state=i).median()
    for i in range(2000)
])
boot.quantile([0.025, 0.975])
```

```python
boot.plot.hist(bins=40, title="2,000 bootstrap medians, Riverside");
```

```python
from scipy import stats

result = stats.bootstrap(
    (riverside,), np.median, n_resamples=2000, method="percentile", rng=1460
)
result.confidence_interval
```
