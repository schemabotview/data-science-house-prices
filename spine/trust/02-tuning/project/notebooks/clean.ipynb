```python
q1, q3 = homes["price"].quantile([0.25, 0.75])
low, high = q1 - 1.5 * (q3 - q1), q3 + 1.5 * (q3 - q1)
low, high, (homes["price"] < low).sum(), (homes["price"] > high).sum()
```

```python
odd = (homes["price"] < low) | (homes["beds"] > 6)
homes.loc[odd, ["id", "agent", "sqft", "beds", "price"]]
```

```python
homes.loc[homes["beds"] == 12, "beds"] = 2
homes = homes[homes["price"] > 20_000]
homes.shape
```

```python
def clean(raw: pd.DataFrame) -> pd.DataFrame:
    """Millbrook's sales, cleaned: every rule from this notebook, in one place."""
    iso = pd.to_datetime(raw["sold"], format="%Y-%m-%d", errors="coerce")
    sold = iso.fillna(pd.to_datetime(raw["sold"], format="%d/%m/%Y", errors="coerce"))
    in_m2 = (raw["agent"] == "Keyline") & sold.between("2023-04-01", "2023-06-30")
    sqft = raw["sqft"].where(~in_m2, (raw["sqft"] * 10.764).round())
    return (
        raw.assign(
            sold=sold,
            area=raw["area"].str.strip(),
            garage=raw["garage"].fillna(0).astype(int),
            price=pd.to_numeric(raw["price"].str.replace("[£,]", "", regex=True)),
            sqft=sqft.astype(int),
            beds=raw["beds"].replace(12, 2),
        )
        .drop_duplicates()
        .query("price > 20_000")
    )
```

```python
raw = pd.read_csv("../data/homes.csv")
tidy = clean(raw)
assert tidy["price"].dtype == "int64" and tidy["sold"].notna().all()
assert not tidy.duplicated().any() and tidy["beds"].max() <= 6
len(raw), len(tidy)
```

```python
tidy.to_csv("../data/homes_clean.csv", index=False)
tidy.dtypes
```
