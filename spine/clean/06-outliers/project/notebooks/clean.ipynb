```python
import pandas as pd

homes = pd.read_csv("../data/homes.csv")
homes.isna().sum()
```

```python
homes.isna().mean().round(3)
```

```python
homes["garage"].value_counts(dropna=False)
```

```python
homes["garage"] = homes["garage"].fillna(0).astype(int)
homes.isna().sum()[lambda s: s > 0]
```

```python
as_number = pd.to_numeric(homes["price"], errors="coerce")
homes.loc[as_number.isna(), ["agent", "price"]].head(4)
```

```python
as_number.isna().sum()
```

```python
digits = homes["price"].str.replace("£", "").str.replace(",", "")
homes["price"] = pd.to_numeric(digits)
homes["price"].dtype, homes["price"].median()
```

```python
iso = pd.to_datetime(homes["sold"], format="%Y-%m-%d", errors="coerce")
uk = pd.to_datetime(homes["sold"], format="%d/%m/%Y", errors="coerce")
homes["sold"] = iso.fillna(uk)
per_bed = homes["sqft"] / homes["beds"]
per_bed.describe().round()
```

```python
homes.loc[per_bed < 200, ["sold", "agent", "sqft", "beds"]].head(4)
```

```python
spring_2023 = homes["sold"].between("2023-04-01", "2023-06-30")
batch = (homes["agent"] == "Keyline") & spring_2023
batch.sum(), (per_bed[batch] < 200).sum(), (per_bed[~batch] < 200).sum()
```

```python
homes.loc[batch, "sqft"] = (homes.loc[batch, "sqft"] * 10.764).round().astype(int)
(homes["sqft"] / homes["beds"]).describe().round()
```

```python
homes.duplicated().sum()
```

```python
homes[homes.duplicated(keep=False)][["id", "sold", "area", "price"]].head(4)
```

```python
homes = homes.drop_duplicates()
homes.shape
```

```python
homes[["beds", "price"]].describe().round()
```

```python
q1, q3 = homes["price"].quantile([0.25, 0.75])
low, high = q1 - 1.5 * (q3 - q1), q3 + 1.5 * (q3 - q1)
low, high, (homes["price"] < low).sum(), (homes["price"] > high).sum()
```

```python
odd = (homes["price"] < low) | (homes["beds"] > 6)
homes.loc[odd, ["id", "agent", "sqft", "beds", "price"]]
```

```python
homes.loc[homes["beds"] == 12, "beds"] = 2
homes = homes[homes["price"] > 20_000]
homes.shape
```
